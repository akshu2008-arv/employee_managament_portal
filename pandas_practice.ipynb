## 0) Setup

```python
import os
import pandas as pd

path = os.path.join("data", "employees.csv")
df = pd.read_csv(path)
df
```

## 1) Explore the table

```python
df.head()          # first 5 rows
```

```python
df.tail()          # last 5 rows
```

```python
df.shape           # (rows, columns)
```

```python
df.columns         # column names
```

```python
df.info()          # types + nulls
```

```python
df.describe(include="all")
```

## 2) Select columns

```python
df["name"]
```

```python
df[["employee_id", "name", "department", "salary"]]
```

## 3) Fix types (important for salary filters)

```python
df["salary"] = pd.to_numeric(df["salary"], errors="coerce")
df["is_active"] = df["is_active"].astype(str).str.lower().map({"true": True, "false": False})
df.dtypes
```

## 4) Filter rows (WHERE in SQL)

```python
df[df["department"].str.lower() == "hr"]
```

```python
df[df["is_active"] == True]
```

```python
df[df["salary"] > 60000]
```

```python
df[(df["department"].str.lower() == "hr") & (df["salary"] > 60000)]
```

```python
df[df["name"].str.contains("a", case=False, na=False)]
```

## 5) Sort

```python
df.sort_values("salary", ascending=False)
```

```python
df.sort_values(["department", "salary"], ascending=[True, False])
```

## 6) Add / change columns

```python
df["bonus"] = df["salary"] * 0.1
df[["name", "salary", "bonus"]]
```

```python
df["department_upper"] = df["department"].str.upper()
df[["department", "department_upper"]]
```

## 7) Group by (like SQL GROUP BY)

```python
df.groupby("department")["salary"].mean()
```

```python
df.groupby("department").agg(
    count=("employee_id", "count"),
    avg_salary=("salary", "mean"),
    max_salary=("salary", "max"),
)
```

```python
df["is_active"].value_counts()
```

## 8) Missing values

```python
df.isna().sum()
```

```python
# example: fill missing salary with 0 (practice only)
df["salary"] = df["salary"].fillna(0)
```

## 9) Pick rows by position / id

```python
df.iloc[0]          # first row
```

```python
df.loc[df["employee_id"] == "EMP002"]
```

## 10) Save (do NOT overwrite original while practicing)

```python
out_path = os.path.join("data", "employees_practice.csv")
df.to_csv(out_path, index=False)
print("Saved to", out_path)
```
